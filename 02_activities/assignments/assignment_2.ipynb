  ## Examples

  ### Example 1

  Input: `lst = [0, 2]`

  Output: [1]

  ### Example 2

  Input: `lst = [5, 0, 1]`

  ### Example 3

  Input: `lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]`

  Output: [4, 9]

#### Starter Code for Question 3


```python
from typing import List

def missing_num(nums: List[int]) -> List[int]:
    n = max(nums)  
    full_set = set(range(n + 1))
    missing = sorted(list(full_set - set(nums)))
    return missing
```

```python
#We’re given a string of brackets that may contain only (), {}, and [] (and possibly spaces). The question asks whether the string is “valid”: every opening bracket must be closed by the same type, and the pairs must be properly nested and ordered. If all brackets match correctly and nothing is left unclosed, the string is valid; otherwise, it’s invalid.
```


-   Create 1 new example that demonstrates you understand the problem. Trace/walkthrough 1 example that your partner made and explain it.


```python
# New example (mine):

Input: s = "([{}])"

Expected Output: True

Why: The nesting is correct: ( opens, then [, then {; they close in reverse order }, ], ).

Trace/walkthrough of my partner’s example:

Partner’s example: s = "([)]"

Claimed Output: False
Step-by-step:

Read '(' → push '(' → stack: ['(']

Read '[' → push '[' → stack: ['(', '[']

Read ')' → expects top '(', but top is '[' → mismatch → return False.
This shows the order is wrong (they crossed), so the string is invalid.
```


-   Copy the solution your partner wrote. 


```python
# Your answer here

def is_valid_brackets(s: str) -> bool:
    # Return True if s is a valid bracket sequence.
    pairs = {')': '(', ']': '[', '}': '{'}
    openers = set(pairs.values())
    stack = []

    for ch in s:
        if ch in openers:
            stack.append(ch)
        elif ch in pairs:          # closing
            if not stack or stack.pop() != pairs[ch]:
                return False
        elif ch.isspace():         # ignore spaces if present
            continue
        else:                      # any other char makes it invalid
            return False

    return not stack
```


-   Explain why their solution works in your own words.


```python
#Their solution uses a stack, which naturally models nested structures. Every time an opening bracket appears, it’s pushed onto the stack. When a closing bracket appears, the code checks that the most recent opener (the top of the stack) is the matching type; if not, it’s invalid. If all closers properly match and the stack is empty at the end, the sequence is valid. They also ignore spaces and reject any other characters, which makes their function robust to minor formatting.
```


-   Explain the problem’s time and space complexity in your own words.


```python
#Time Complexity: O(n), because each character is processed once (push or pop is O(1)).
#Space Complexity: O(n) in the worst case, when the string consists of only opening brackets (they all sit on the stack until matching closers arrive).
```
